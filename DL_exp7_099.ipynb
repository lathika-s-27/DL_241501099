```python
import requests

data = requests.get("https://dog.ceo/api/breed/labrador/images/random").json()
url = data["message"]

image = requests.get(url).content

with open("labrador.jpg", "wb") as f:
    f.write(image)

print("Labrador image downloaded successfully")
```
